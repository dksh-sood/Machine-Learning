```python
## series 
import pandas as pd
data =[1,2,3,4]
series = pd.Series(data)
print("series \n",series)
```

```python
## ceate a series from dictionary elements 
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index =['a','b','c']
pd.Series(data,index=index)
```

```python
##dataframe --> having multiple row and col
## create a dataframe from a dictionay of list

data = {
  'Name':['daksh','shruti','karan'],
  'Age':[21,20,5],
  'City':['Delhi','Nanded','Pune']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python

data = [
  {'Name' : 'Daksh', 'Age': 21 ,'City': 'Banglore'},
  {'Name' : 'shruti', 'Age': 20 ,'City': 'Nanded'},
  {'Name' : 'Shashank', 'Age': 20 ,'City': 'Rajasthan'},
  {'Name' : 'Shagun', 'Age': 19 ,'City': 'Meerut'}
]
df = pd.DataFrame(data)
print(df)
```

```python
### accessing data from data frame
df['Name']
```

```python
## .loc --> label based selection (selecting data using row or column names (labels))
df.loc[0][0] # giving warning
```

```python
## index based selection (u selecting data using row/column position(number))
df.iloc[0][0]
```

```python
## Accessing a speciified element
df.at[2,'Age']
```

```python
df.at[2,'Name']
```

```python
## accessing element using iat
df.iat[2,2]
```

```python
## data manipulation with df
df['salary']=[50000,60000,70000,10000]
df
```
