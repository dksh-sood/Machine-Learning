```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y=[1,4,9,16,25]

# create a line plot 
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title("Basic Line Plot")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

## create a customized line plot

plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth=2,markersize=6)
plt.grid(True)
```

```python
## multiple plots
## sample data

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1) #plt.subplot(nrows, ncols, index)
plt.plot(x,y1,color='green')
plt.title("Plot 1")


plt.subplot(2,2,2) #plt.subplot(nrows, ncols, index)
plt.plot(y1,x,color='blue')
plt.title("Plot 2")

plt.subplot(2,2,3) #plt.subplot(nrows, ncols, index)
plt.plot(x,y2,color='red')
plt.title("Plot 3")

plt.subplot(2,2,4) #plt.subplot(nrows, ncols, index)
plt.plot(x,y2,color='orange')
plt.title("Plot 4")

plt.tight_layout()
plt.show()
```

```python
### bar plot --> plot related to categories 
categories = ['A','B','C','D','E']
values = [5,7,3,8,6]

## create a bar plot
plt.bar(categories,values,color='purple')
plt.xlabel('categories')
plt.ylabel('values')
plt.title("Bar Plot")
plt.show()
```

### Histogram -> represent the distributionnof dataset. they divide the data ontp bins and count the number of data points in each bin

```python
data = [1,2,2,3,3,3,4,4,4,5,5,5,5]

# create a histogram 
plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
## create a scatter plot 
## sample data
x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.scatter(x,y,color='blue',marker='x')
```

```python
## pie chart

labels = ['A','B','C','D']
sizes = [30,20,40,10]
colors = ['gold','yellowgreen','lightcoral','lightskyblue']
explode = (0.2,0,0,0) # move out the 1st slice

## create apie chart
plt.pie(sizes,explode = explode,labels=labels,colors=colors,autopct="%1.1f%%",shadow=True)
```

```python
## plot total sales by product
total_sales_by_product = sales_data_df.groupby('Product Category')['Total Revenue'].sum()
print(total_sales_by_product)
```
