## Arbitary Argument

```python
def my_func(*kids):
  print("the youngest child  " +kids[2])
my_func("Daksh","Sood","Harsh")
```

## Lambda Function

```python
number = [1,2,3,4,5]
square = list(map(lambda x: x**2, number))
print(square)
square_root = list(filter(lambda x: x%2==0,number))
print(square_root)
```
