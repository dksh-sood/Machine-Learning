```python
import matplotlib.pyplot as plt

x=[10,20,30,40]
y=["c","c++","java","python"]
ex=[0.4,0.0,0.0,0.0]
c=["r","b","g","m"]
plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.1f%%",shadow=True,radius=1.1,labeldistance=1.05,startangle=190
        ,textprops={"fontsize":20},counterclock=False,wedgeprops={"linewidth":5,"edgecolor":"pink"}
        ,)#rotatelabels=5,,wedgeprops={"linewidth":4}-->increase the shadow width,,centere(3,3)
plt.title("Daksh Sood")
plt.legend(loc="upper left")
plt.show()



```
