```python
import matplotlib.pyplot as plt 
import numpy as np 

x = [1,2,3,4]
y = [2,4,6,8]

c=["r","y","g","b"]
plt.bar(x,y,color=c) 
plt.show()
```

```python
import matplotlib.pyplot as plt 
import numpy as np 

x =["python","c","c++","java"]
y=[85,70,60,82]
plt.xlabel("language",fontsize = 30)
plt.ylabel("number",fontsize = 30)
plt.title("daksh sood",fontsize = 30)
c=['r','b','m','y']
plt.bar(x,y,width=0.2,color=c,align="center",edgecolor = "black", linewidth=2, linestyle=":",alpha = 0.4,label = "daksh")# align edge // alpha is for light the color
plt.legend(loc = 'upper center')
plt.show()
```

```python
import matplotlib.pyplot as plt 
import numpy as np 

x =["python","c","c++","java"]
y=[85,70,60,82]
z=[20,30,40,50]

width = 0.2
p=np.arange(len(x))
p1=[j + width for j in p]

plt.xlabel("language",fontsize = 15)
plt.ylabel("number",fontsize = 15)
plt.title("daksh sood",fontsize = 15)

plt.bar(p,y,width,color="r",edgecolor = "black", linewidth=2, linestyle=":",alpha = 0.4,label = "daksh")# align edge // alpha is for light the color(0---1)
plt.bar(p1,z,width,color="y",label = "sood",linestyle=":",edgecolor ="black") #align center use nhi krna hota fir dusra shade use nhi dikhta

plt.xticks(p+width/2,x,rotation = 10)#pyhton , java bich m show hoga 
plt.legend(loc = 'upper center')
plt.show() 
```
