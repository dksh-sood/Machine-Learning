```python
import matplotlib.pyplot as plt


day = [1,2,3,4,5,6,7]
no = [2,3,1,4,5,3,6]

plt.scatter(day,no,color="r")
plt.title("daksh sood",fontsize=30)
plt.xlabel("day",fontsize=30)
plt.ylabel("No",fontsize=30)
plt.show()
```

```python
import matplotlib.pyplot as plt


day = [1,2,3,4,5,6,7]
no = [2,3,1,4,5,3,6]
color =["r","y","g","b","r","g","r"]
sizes=[400,200,400,300,200,300,156]
plt.scatter(day,no,c = color,s=sizes,alpha =0.4,marker="*",edgecolor="black",linewidth=4)
plt.title("daksh sood",fontsize=30)
plt.xlabel("day",fontsize=30)
plt.ylabel("No",fontsize=30)
plt.show()
```

```python
import matplotlib.pyplot as plt
day = [1,2,3,4,5,6,7]
no =  [2,3,1,4,5,3,6]
no2 = [3,2,4,5,1,6,2]
color =[10,23,41,65,89,78,63]
sizes=[400,200,400,300,200,300,156]
plt.scatter(day,no,c=color,s=sizes,marker="*",cmap="viridis")
plt.scatter(day,no2,color="r",s=sizes)
t=plt.colorbar()
t.set_label("colorbar",fontsize=30)
plt.title("daksh sood",fontsize=30)
plt.xlabel("day",fontsize=30)
plt.ylabel("No",fontsize=30)
plt.show()
```
