```python
import pandas as pd
import numpy as np

data = {
    'Age':np.random.randint(20 ,60 ,size = 1000),
    'Salary':np.random.randint(10000,100000,size = 1000),
    'Department':np.random.choice(['IT department','Finance department','Mechanical department','Survey department'],size = 1000),
    'Expenses':np.random.randint(500,5000,size=100),
    'Weight':np.random.randint(50,120,size = 1000),
    'height':np.random.randint(150,200,size = 1000),
    'Disease':np.random.choice(['yes','no'],size=1000)
}
df = pd.DataFrame(data)
df.to_csv('info.csv',index=False)
```
