```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('Housing.csv')


print(data.head()) 


plt.figure(figsize=(10, 6))
plt.bar(data['area'], data['price'], color='r',linewidth=2,width=4,label="price/area")
plt.title('Flat Area vs Price',fontsize=30)
plt.xlabel('Area (sq meters)',fontsize=30)
plt.ylabel('Price-per square feet',fontsize=30)
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('Housing.csv')


print(data.head()) 


plt.figure(figsize=(10, 6))
plt.scatter(data['price'],data['area'], color='r')
plt.title('Flat Area vs Price',fontsize=30)
plt.xlabel('Area (sq meters)',fontsize=30)
plt.ylabel('Price-per square feet',fontsize=30)
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('Housing.csv',nrows=100)


print(data.head()) 


plt.figure(figsize=(10, 6))
plt.bar(data['area'], data['price'], color='r',linewidth=2,width=10,label="price/area")
plt.title('Flat Area vs Price',fontsize=20)
plt.xlabel('Area (sq meters)',fontsize=20)
plt.ylabel('Price-per square feet',fontsize=20)
plt.legend()
plt.grid()
plt.show()
```
