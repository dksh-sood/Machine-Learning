```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],"B":[11,22,33,44]},index=["a","b","c","d"])
var2 = pd.DataFrame({"C":[10,40],"D":[43,54]},index=["a","b"]) #index is alway equal to the given value
var2.join(var1,how="right") #we can inter change the var1<-->var2
#how =(left,right,outer-->{union ka kaam krta h},inner-->{inter section data})
```
