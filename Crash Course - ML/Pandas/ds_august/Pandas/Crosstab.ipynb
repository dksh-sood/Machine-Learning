```python
import pandas as pd
```

```python
df = pd.read_csv("Test_new3.csv")
df
```

Crosstab function --> also known as the contingency table and cross tabulation
--> it a type of table in a matrix format that is used for count thefrequency of the variable & the element

-->crosstab method is only on the pandas and not on the dataframes

```python
pd.crosstab([df.Sex,df.Handedness],[df.Nationality],margins=True) #margins is used for getting total no. od handedness
```

```python
pd.crosstab([df.Sex],[df.Handedness],normalize='index') #normalize is giving na percentage
```

```python
import numpy as np
pd.crosstab([df.Sex],[df.Handedness],values=df.Age,aggfunc=np.average)
```
