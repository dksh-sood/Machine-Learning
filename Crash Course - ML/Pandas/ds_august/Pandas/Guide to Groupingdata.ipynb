GROUPBY-->

```python
var = pd.DataFrame({"Name":["e","d","a","f","c","b","a","e","f"],
                    "sub_1":[12,13,14,15,16,17,18,19,20],
                    "sub_2":[22,23,24,25,26,27,28,29,30],
                    "sub_3":[32,33,34,35,36,37,38,39,40],
                    "sub_4":[42,43,44,45,46,47,48,49,50]})
var
```

```python
#group of same kind of data
var_New = var.groupby("Name")
var_New
for x,y in var_New: #used for printing!!
    print(x)
    print(y)
    print() #to get the space
```

```python
var_New.get_group("a") # to get specific same kind of group
```

```python
var_New.min()
var_New.max()
var_New.mean()
```
