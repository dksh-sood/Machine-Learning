```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_excel('Sales Analysis Report.xlsx')
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

year_data =df[df['OrderDate'].dt.year==2018]
Year_sales_trend = year_data.groupby('OrderDate')['Total Revenue'].sum()
month_sales = year_data.groupby(year_data['OrderDate'].dt.month)['Total Revenue'].sum()
```

```python
plt.Figure(figsize=(10,6))
c=['r','g','b','m','g','y','b','r','g','b','m','r']
plt.bar(month_sales.index,month_sales.values , color=c,alpha=0.4,width=0.6,linestyle=":",linewidth=2)+
plt.title('Sales Trend')
plt.xlabel('order date')
plt.ylabel('Total Revenue')
plt.grid()
plt.xticks(month_sales.index,['jan','feb','mar','apr','may','jun','jul','aug','sept','oct','nov','dec'])
plt.show()
```
