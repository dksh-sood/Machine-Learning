```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],"B":[11,22,33,44]})

var2 = pd.DataFrame({"C":[12,32,43,54],"B":[11,22,33,48]})

pd.merge(var1,var2,on="B") #interchnage according to our choice
```

var1 and var2 -->  are merge when they have 1 common element ,, if their is any uncommon number to wha tk print hoga jha tk common hoga !!! 

```python
pd.merge(var1,var2,how="outer",indicator=True)
```

how="(outer/right/left)"

```python
pd.merge(var1,var2,left_index=True,right_index=True,suffixes=("name","id")) #using index we can see any data
```

CONCAT in python pandas
merging of same heading in the one row 

```python
#concat of two series
sr1=pd.Series([1,2,3,4])
sr2=pd.Series([11,22,33,44])

pd.concat([sr1,sr2])
```

```python
#concat of two data-frames
d1 = pd.DataFrame({"A":[1,2,3,4],"B":[11,22,33,44]})
d2 = pd.DataFrame({"A":[12,32],"B":[11,48]}) #dono ka same size ka hona zaruri hai

pd.concat([d1,d2],axis=1,join="outer",keys=["d1","d2"]) #axis = 1 (concat along the col)
```
