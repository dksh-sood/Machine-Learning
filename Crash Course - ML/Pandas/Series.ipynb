```python
import pandas as pd
#Series data str
x = [3,4,5,6,7,8] #use list,tuple,dictionary
var = pd.Series(x,index=['a','b','c','d','e','f'],dtype="float"
                ,name="python")
print(var)
print(type(var))
print(var[2]) #to get any data or num
```

```python
#series through dictionary
dict ={"name":['python','c','c++','java'],"por":[12,13,14,15],"rank":[1,2,3,4],}
var = pd.Series(dict)
print(var)
```

```python
#adding same data by giving index value
s = pd.Series(12,index=[1,2,3,4,5,6,7])
print(s)
```

```python
#additon of two series element
s1 = pd.Series(12,index=[1,2,3,4,5,6,7])
s2 = pd.Series(12,index=[1,2,3,4])#numpy give broadcast error

print(s1 + s2)
```
