
-->data arrangment and reshaping

```python
var = pd.DataFrame({"Days":[1,1,1,1,2,2],
                    "student_Name":["daksh","karan","harshu","khushi","sunaina","nitish"]
                    ,"Eng":[10,20,30,40,50,60],"Maths":[12,14,15,16,17,18]})
var
```

```python
var.pivot(index="Days",columns="student_Name") #two arrgumnet r compulsary
```

```python
var.pivot_table(index="student_Name",columns="Days",aggfunc = "mean") #mean , count , max , sum , min 
```

```python
var.pivot_table(index="student_Name",columns="Days",aggfunc = "sum",margins=True)
```
