```python
import pandas as pd
#insert
var = pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10],"C":[1,2,3,4,5],"D":[6,7,8,9,10]})
print(var)
var.insert(1,"python",var["A"])#(index no./position,name,kya data isert karana hai)
print(var)
var.insert(2,"python_1",[11,22,33,44,55])
print(var)
```

```python
var["python_12"] = var["A"][:3] #copy and slicing
print(var)
```
