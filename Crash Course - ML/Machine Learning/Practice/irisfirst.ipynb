## Data Analysis using Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion='entropy',random_state=42)
classifier.fit(X_train,Y_train)
```

```python
y_pred = classifier.predict(X_train)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
cm = confusion_matrix(Y_train,y_pred)
print(cm)
accuracy_score(Y_train,y_pred)
```

```python
def encrypt(text, key):
  """Encrypts a text using a Caesar cipher.

  Args:
    text: The text to encrypt.
    key: The encryption key.

  Returns:
    The encrypted text.
  """
  result = ''
  for char in text:
    if char.isalpha():
      start = ord('a') if char.islower() else ord('A')
      shifted_char = chr((ord(char) - start + key) % 26 + start)
    elif char.isdigit():
      shifted_char = str((int(char) + key) % 10)
    else:
      shifted_char = char
    result += shifted_char
  return result


def decrypt(text, key):
  """Decrypts a text encrypted with a Caesar cipher.

  Args:
    text: The text to decrypt.
    key: The encryption key.

  Returns:
    The decrypted text.
  """
  return encrypt(text, -key)


text = input("Enter the text to encrypt: ")
key = int(input("Enter the encryption key: "))

encrypted_text = encrypt(text, key)
print("Encrypted text:", encrypted_text)

decrypted_text = decrypt(encrypted_text, key)
print("Decrypted text:", decrypted_text)
```
