```python
def count_even_odd(lst):
    even_count = 0
    odd_count = 0
    for i in lst:
        if i%2==0:
            even_count +=1
            return even_count
        else:
            odd_count+=1
            return odd_count
lst = [1,2,3,4,5,6]
print(count_even_odd(lst))
```

```python
def max_consecutive_difference(lst):
  high = lst[0]
  for i in lst:
    if lst[i] < high:
      high = lst[i]
  return high
  
  lst = [1,2,3,4,5]
  print(max_consecutive_difference(lst))
```
