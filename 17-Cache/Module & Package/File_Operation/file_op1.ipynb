## file operation - read and write files

```python
# read whole file
with open('example.txt','r') as file:
  content = file.read()
  print(content)
```

```python
# read file line by line
with open('example.txt','r')as file:
  for line in file:
    print(line.strip()) # remove new line char
```

```python
#writing a file(overwriting)
with open('example.txt','w') as file:
  file.write('hellow world!\n')
  file.write("this is a new line\n")
```

```python
#write a file(without overwriting)
with open('example.txt','a') as file:
  file.write("append operation taking place!\n")
```

```python
## writing a list of lines to file
lines = ['first line \n','second line\n','third line\n']
with open('example.txt','a') as file:
  file.writelines(lines)
```

```python
## binary files
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
  file.write(data)
```

```python
## binary files
with open('example.bin','rb') as file:
  content = file.read()
  print(content)
```

```python
# read the content form source text file and write to a destination text file
# copy a text file
with open('example.txt','r') as source_file:
  content =source_file.read()
with open('example.txt','w') as destination_file:
  destination_file.write(content)
```

```python
# writing and then reading file

with open('example.txt','w+') as file:
  file.write("hello world\n")
  file.write("this is a new line \n")

  # move the cursor to the beginning
  file.seek(0)

  ## read the content of the file
  content = file.read()
  print(content)
```

```python
## create a new directory
import os
new_directory = "package"
os.mkdir(new_directory)
print(f"directory '{new_directory}' create")
```

```python
# listing files and dir
items = os.listdir('.') # . give currently working location
print(items) # give all files in this folder
```

```python
#joining paths
dir_name = "folder"
file_name="file.txt"
full_path = os.path.join(dir_name,file_name)
print(full_path)
```

```python
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd(),dir_name, file_name)
print(full_path)
```

```python
# to check file is there or not 
path="example.txt"
if os.path.exists(path):
  print(f"the path '{path}' exists")
else:
  print(f"the path '{path}' does not exists")
```

```python
#checking is a path is a file or dir
import os

path = "example.txt"
if os.path.isfile(path):
  print(f"the path '{path}' is a file")
elif os.path.isdir(path):
  print(f"the path '{path}' is directory")
else:
  print(f"the path '{path}' is neither a file nor a directory")
```

```python
# getting absolute path
relative_path = 'example.txt'
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
