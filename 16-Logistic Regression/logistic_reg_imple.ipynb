## Logistic Regression Implementation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification  # create dataset like balanced data and donot need to standardise the  dataset
```

```python
## create dataset
X,y = make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
pd.DataFrame(X).head(5)
```

```python
pd.DataFrame(y).head(5)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=30,random_state=42)
```

```python
## Model Training
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression() 
```

```python
logistic.fit(X_train,y_train)
```

```python
y_pred = logistic.predict(X_test)
print(y_pred)
```

```python
# logistic.predict_proba(X_test)  --> used yo check the prob of point with upper and lower class from bestfit line 
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
score = accuracy_score(y_test,y_pred)
print("accuracy score is: ",score)
cm = confusion_matrix(y_test,y_pred)
print("confusion metrics is :",cm)
print("classification_report is :\n",classification_report( y_test ,y_pred))
```

```python
## accuracy score is:  0.7666666666666667   (before hyper-parameter tunning and cros-validation)
```

## Hyper-Parameter tunning and Cross Validation

```python
## for custome parameters or use another parameters we use hyperparamenter tunning(for best attribute selection and doesnot leads to overfitting)
model = LogisticRegression()
penalty=['l1','l2','elasticnet'] # to check best parameters
c_Values = [100,10,1.0,0.01]
solver = ['lbfgs', 'liblinear','newton-cg', 'newton-cholesky', 'sag', 'saga']
```

```python
## Coverting into key value pair
params = dict(penalty = penalty,C = c_Values,solver=solver)
```

## Grid-SearchCV

```python
## Grid-SearchCV
from sklearn.model_selection import GridSearchCV , StratifiedKFold
cv = StratifiedKFold()
grid = GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_  ## best combination is selected there 
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print("accuracy score is: ",score)
cm = confusion_matrix(y_test,y_pred)
print("confusion metrics is :",cm)
print("classification_report is :\n",classification_report( y_test ,y_pred))
```

```python
## accuracy score is:  {0.8} after the hyper-parameter tunning and cross validation

## GridSearchCV = “Har possible parameter try karo + cross validation karo + best select karo”
```

| Parameter  | Meaning                    |
| ---------- | -------------------------- |
| estimator  | Kaunsa model               |
| param_grid | Kaunse parameters try kare |
| scoring    | Accuracy se measure kare   |
| cv         | Cross validation method    |
| n_jobs=-1  | Saare CPU cores use kare   |


## Randomized Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
randomcv = RandomizedSearchCV(estimator=model,param_distributions=params , cv = 5, scoring='accuracy')
```

```python
randomcv.fit(X_train,y_train)
```

```python
print(randomcv.best_score_)
```

```python
randomcv.best_params_
```

```python
y_pred = randomcv.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print("accuracy score is: ",score)
cm = confusion_matrix(y_test,y_pred)
print("confusion metrics is :",cm)
print("classification_report is :\n",classification_report( y_test ,y_pred))
```

```python
# why we use randomcv becuz when we have large number of parameters it take less number of time compare to GridSearchCV
```

## Logistic Regression for Multiclass classification Problem

```python
## create dataset
X,y = make_classification(n_samples=1000,n_features=10,n_informative=3,n_classes=3,random_state=42)
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression(multi_class='ovr')
logistic.fit(X_train,y_train)
y_pred=logistic.predict(X_test)
```

```python
y_pred
```

```python
score = accuracy_score(y_pred,y_test)
print("accuracy score is: ",score)
cm = confusion_matrix(y_pred,y_test)
print("confusion metrics is :",cm)
print("classification_report is :\n",classification_report(y_pred,y_test))
```
