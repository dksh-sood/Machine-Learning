```python
##5
def word_count_file(file_path):
  with open(file_path,'r') as file:
    lines = file.readlines()
    word_count = sum(len(line.split()) for line in lines)
  return word_count

file_path = 'log.txt'
word = word_count_file(file_path)
print(f'the word count is : {word}')

```

```python
##6
def find_and_count(filename,old_word,new_word):
  with open(filename , 'r') as file:
    text = file.read()
  new_text = text.replace(old_word,new_word)
  with open(filename,'w') as file:
    file.write(new_text)
  with open(filename, 'r') as file:
    content = file.read()
  print(content)
find_and_count('data.txt','daksh','shruti')

```

```python
##7
def reverse_word(filename):
  with open(filename,'r') as file:
    lines = file.readlines()
  for line in reversed(lines):
    print(line.strip())
reverse_word('data.txt')
```

```python
##8
def word_stats(filename):
  with open(filename , 'r') as file:
    lines = file.readlines()
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)
  return line_count , word_count , char_count

filename = 'data.txt'
line , word , char = word_stats(filename)
print(f'number of line is : {line} and the number of words is : {word} and the number of charachter is : {char}')
```

```python
##9
file = open('merged.txt','w')
def merging_file(input_file , output_file):
  with open(output_file , 'w') as outfile:
    for mer in input_file:
      with open(mer,'r') as infile:
        outfile.write(infile.read())
merging_file(['data.txt','sample.txt','output.txt','log.txt'],'merged.txt')
```

```python
##11
import datetime
def creating_log(message , filename='activity.log'):
  timestamp = datetime.datetime.now().isoformat()
  with open(filename, 'a') as file:
    file.write(f'[{timestamp}] , {message}\n')
creating_log('hey this is a log file')
```

```python
##12
def binary_operation(source , destination):
  with open(source , 'rb') as src:
    with open(destination , 'wb') as dest:
      dest.wwrite(src.read())
binary_operation()
```

```python
##13
import csv
def csv_operation(filename):
  with open(filename , 'r') as file:
    reader = csv.DictReader(file)
    return list(reader)

print(csv_operation('data.csv'))
```

```python
##14
import json
def json_operation(filename):
  with open(filename,'r') as file:
    data = json.load(file)
    return list(data)
print(json_operation('data.json'))
```

```python
##15
def read_protected(file):
  try:
    with open(filename,'r') as file:
      file.read()
  except PermissionError as e:
    print(f'Permission Erro {e}')

read_protected('protected.txt')
```
