# Logging:
Logging is a crucial aspect of any application, providing a way to track events, errors and operational information. Python built in logging module offers a flexible framework for emitting log messages from python progrms. In this lesson, we will cover the basic of logging, including how to configure logging  log levels and best practices for using logging in python application

```python
import logging

## configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

## log messages with different severity level
logging.debug("this is a debug message")
logging.info("this is an info message")
logging.warning("this is a warning message")
logging.error("this is an error meassage")
logging.critical("this is a critical message")
```

# Log Level
### DEBUG : detailed information, typically of interest only when diagnosing problems
### INFO : confirmation that things are working as expected
### WARNING : an indication that something unexpected happened or indicative of some problem in the near future(eg disk space low) . the software is still working as expected
### ERROR: Due to a more serious problem, the software has not been able to perform some function
### CRITICAL: A very serious error, indicating that the program itself may be unable to continue running

```python
## configuring logging
import logging

logging.basicConfig(
  filename='app.log',
  filemode='w',
  level=logging.DEBUG,
  format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',  
  datefmt='%Y-%m-%d %H:%M:%S',
  force = True  # this reset existing config
)

## log messages with different severity levels
logging.debug("this is a debug message")
logging.info("this is an info message")
logging.warning("this is a warning message")
logging.error("this is an error message")
logging.critical("this is a critical message")
```

# Logging with multiple loggers

you can create multiple loggers for a different parts of your application

```python
import logging
## create a logger for module1

logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)

## create a logger for module 2

logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)

## configure logging settings
logging.basicConfig(
  level=logging.DEBUG,
  format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',  
  datefmt='%Y-%m-%d %H:%M:%S',
  force = True  # this reset existing config
)
```

```python
## log messages with different loggers
logger1.debug("this is a debug message for module1")
logger2.warning("this is a warning message for module 2")
logger2.error("this is an error message")
```
