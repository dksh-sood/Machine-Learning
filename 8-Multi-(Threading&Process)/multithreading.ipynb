# Multithreading
#### 1. when to use multi threading ?
####  i/o bound tasks : tasks that spend more time waiting for i/o operations (eg-> file operations , network requests)
#### * concurrent execution: when you want to improve the throughput of your application by performing multiple operations concurrently.
###### example :
 ###### Imagine you’re cooking 🍳

###### You put rice on the stove.
###### While rice is boiling, you cut vegetables.
###### Then you stir the rice again.
###### Then you make salad.
###### You’re not truly doing both at the same exact moment,
but you’re switching between tasks so fast that it feels like multitasking.

```python
import threading
import time

def print_numbers():
  for i in range(5):
    time.sleep(.5)
    print(f"Number:{i}")

def print_letter():
  for letter in "abcde":
    time.sleep(.5)
    print(f"letter: {letter}")

## create 2 threads
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letter)

#start timer
t=time.time()
#run in parallel
t1.start()
t2.start()

##wait for threads to complete
t1.join()
t2.join()

finished_time = time.time()-t
print(finished_time)
```
