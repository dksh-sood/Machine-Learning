# Generator :
-> it is a simple way to create iterator. They use the `yield` keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
  for i in range(3):
    return i**2
```

```python
square(3) ## iterate all through elements  but first element is 0, but it return 0 like 0 to the times of 2
```

```python
def square(n):
  for i in range(3):
    yield i**2 # save the value and return it !
```

```python
for i in square(3):
  print(i)
```

```python
a = square(3)
a
```

```python
next(a) 
```

```python
def my_generator():
  yield 1
  yield 2
  yield 3
```

```python
gen = my_generator()
```

```python
next(gen)
```

```python
for val in gen:
  print(val)
```

# Practical Example: Reading Large Files

### Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory

```python
## practical : reading large file

def read_large_file(file_path):
  with open(file_path , 'r', encoding='latin-1') as file:
    for line in file:
      yield line
```

```python
file_path = 'large_file.txt'
for line in read_large_file(file_path):
  print(line.strip())
```
