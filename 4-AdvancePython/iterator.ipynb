```python
## create iterator giving number from 1 to 5
class Count:
  def __init__(self):
    self.num = 1
  
  def __iter__(self):
    return self

  def __next__(self):
    if self.num <= 5:
      val = self.num
      self.num +=1
      return val
    else:
      raise StopIteration

for number in Count():
    print(number)
```
