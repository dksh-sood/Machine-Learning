# Decorators
decorators are a powerful and flexible features in python that allows you to modify the behavior of a function or class methods. they are commonly used to add functionality to functions or methods without modifyingtheir actual code. this lesson covers the basics of decorators, including how to create and use them

```python
## function copy
def welcome():
  return "welcome to the advanced python course"

welcome()
```

```python
wel = welcome
print(wel())
del welcome
print(wel())
```

```python
# closures (function inside function)
def main_welcome():
  msg="welcome"
  def sub_welcome_method():
    print("welcome to the advance python course")
    print(msg)
    print("please learn these concepts properly")
  return sub_welcome_method()
```

```python
main_welcome()
```

```python
# closures (function inside function)
def main_welcome(msg):
  
  def sub_welcome_method():
    print("welcome to the advance python course")
    print(msg)
    print("please learn these concepts properly")
  return sub_welcome_method()
```

```python
main_welcome("welcome everyone")
```

```python
# closures (function inside function)
def main_welcome(func):
  
  def sub_welcome_method():
    print("welcome to the advance python course")
    func("i like this tutorial")
    print("please learn these concepts properly")
  return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
# closures (function inside function) -> method inside method , parameter as a methods
def main_welcome(func,lst):
  
  def sub_welcome_method():
    print("welcome to the advance python course")
    print(f'the length of the list is {func(lst)}')
    print("please learn these concepts properly")
  return sub_welcome_method()
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

```python
## Decorators
def main_welcome(func):
  msg="welcome"
  def sub_welcome_method():
    print("welcome to the advance python course")
    func()
    print("please learn these concepts properly")
  return sub_welcome_method()
```

```python
def course_instruction():
  print("this is a advance python course")

course_instruction()
```

```python
main_welcome(course_instruction) # without written main_welcome can i call all the ouput is there ? --> done by decoraters
```

```python
# decorators
@main_welcome # when we write this 
def course_instruction(): # this method is go in main_welcome as a parameter 
  print("this is a advance python course")
```

```python
## Decorator
def my_decorator(func):
  def wrapper():
    print("something is happening before the function called")
    func()
    print("something is happening after the function is called")
  return wrapper
```

```python
@my_decorator
def say_hello():
  print("hello")
```

```python
say_hello()
```

```python
##decorators with arguments
def repeat(n):
  def decorator(func):
    def wrapper(*args,**kwargs):
      for i in range(n):
        func(*args,**kwargs)
    return wrapper
  return decorator
```

```python
@repeat(3)
def say_hello():
  print("hello")
```

```python
say_hello()
```
