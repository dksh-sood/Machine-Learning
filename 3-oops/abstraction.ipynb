## abstraction --> 
hiding the complex implementation and showing only the necessary features of an object, helps to remove the programming complexity and efforts

```python
from abc import ABC,abstractmethod

## abstract class
class Vehicle(ABC):
  def drive(self):
    print("the vehicle is use for driving!")
  
  @abstractmethod
  def start_engine(Self):
    pass

## derived class 1
class Car(Vehicle):
  def start_engine(self):
    print("car engine started")

def operate_vehicle(vehicle):
  vehicle.start_engine()
  vehicle.drive()

car = Car()
operate_vehicle(car)
```
