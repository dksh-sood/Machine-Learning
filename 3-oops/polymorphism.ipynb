## method overriding
method overriding allows child class to provide a specific implementation of method that is already define in its parent class

```python
## base class
class Animal:
  def speak(self):
    return "sound of the animal"

## derived class
class Dog(Animal):
  def speak(self):
    return "woof" # method overriding
## derived class
class Cat(Animal):
  def speak(self):
    return "meow" # method overriding

## function that demonstrate polymorphism
def animal_speak(animal): # Because animal is a parameter (variable) — it represents any object that has a speak() method (Dog, Cat, or Animal).
  print(animal.speak())
# creating object and each object know from which class its belong
dog = Dog() # dog knows it’s an instance of Dog
cat = Cat()
print(dog.speak())  # Dog's method runs
print(cat.speak())  # Cat's method runs
animal_speak(dog) #animal.speak() => dog.speak() → "woof"
```

```python
##polymorphism with function and methods
##base class
class Shape:
  def area(self):
    return "the area of the figure"
# derive class 1
class Rectangle(Shape):
  def __init__(self,width,height):
    self.width = width
    self.height =height
  def  area(self):
    return self.width * self.height
#derived class 2
class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius
  def area(self):
    return 3.14 * self.radius*self.radius
# function that demonstrates ploymorphism

def print_area(shape):
  print(f"the area is {shape.area()}")

rectangle = Rectangle(4,5)
circle = Circle(3)

print_area(rectangle)
print_area(circle)
```

## polymorphism with abstracvts base class

are used to define common methods for a grp or related object. they can enforce that derived class implements particular methods , promoting consistency across different implementation

```python
from abc import ABC, abstractmethod

## define abstract class
class Vehicle(ABC):
  ## abstract method
  @abstractmethod
  def start_engine(self): # method declare not implemented , any class that inherit Vehicle they should write this method
    pass

  ## derived class 1
class Car(Vehicle):
  def start_engine(self):
    return "car engine started"
  
  ## derived class 2
class Motorcycle(Vehicle):
  def start_engine(self):
    return "motorcycle engine started"
## function that demonstartes polymorphism
def start_vehicle(vehicle):
  print(vehicle.start_engine())
    
  ## create object of cae and motorcycle
car = Car()
motorcycle = Motorcycle()
start_vehicle(car)
```
