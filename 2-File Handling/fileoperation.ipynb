# read a text file and count the number of line , words and characters.

```python
def count_txt_file(file_path):
  with open(file_path , 'r') as file:
    lines = file.readlines()
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_cont = sum(len(line) for line in lines)
  return line_count , word_count , char_cont

file_path = 'example.txt'
lines , words , char = count_txt_file(file_path)
print(f'lines: {lines}, words: {words} , charachters: {char}')
```
